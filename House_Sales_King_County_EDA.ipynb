```python
# Máscara de anomalías detectadas

mask_price_zero = df["price"] == 0

mask_bedrooms_bathrooms = (
    (df["bedrooms"] == 0) &
    (df["bathrooms"] == 0)
)

mask_invalid_renovation = (
    (df["yr_renovated"] != 0) &
    (df["yr_renovated"] < df["yr_built"])
)

mask_extreme_price = df.index.isin([4350, 4346])

mask_remove = (
    mask_price_zero |
    mask_bedrooms_bathrooms |
    mask_invalid_renovation |
    mask_extreme_price
)

print("Filas originales:", len(df))
print("Filas a eliminar:", mask_remove.sum())
print("Filas que quedan:", (~mask_remove).sum())
```

```python
df_clean = df.loc[~mask_remove].copy()

print("Filas originales:", len(df))
print("Filas después de la limpieza:", len(df_clean))
print("Filas eliminadas:", len(df) - len(df_clean))
```

```python
# 1. Tamaño final
filas_finales = len(df_clean)
columnas_finales = len(df_clean.columns)

# 2. Calidad de datos
nulos_finales = df_clean.isna().sum().sum()
duplicados_finales = df_clean.duplicated().sum()

# 3. KPIs principales
precio_medio = df_clean["price"].mean()
precio_mediano = df_clean["price"].median()

price_sqft_medio = df_clean["price_per_sqft"].mean()
price_sqft_mediano = df_clean["price_per_sqft"].median()

# 4. Superficie
sqft_living_medio = df_clean["sqft_living"].mean()
sqft_living_mediano = df_clean["sqft_living"].median()

# 5. Waterfront
waterfront_stats = df_clean.groupby("waterfront")["price"].agg(
    ["count", "mean", "median"]
)

# 6. Correlaciones principales con el precio
correlaciones = (
    df_clean.corr(numeric_only=True)["price"]
    .sort_values(ascending=False)
)

# 7. Ciudades con al menos 30 viviendas
city_stats = (
    df_clean.groupby("city")["price"]
    .agg(["count", "mean", "median"])
)

city_stats_30 = (
    city_stats[city_stats["count"] >= 30]
    .sort_values("median", ascending=False)
)

# 8. Precio por sqft por ciudad
city_sqft = (
    df_clean.groupby("city")["price_per_sqft"]
    .agg(["count", "mean", "median"])
)

city_sqft_30 = (
    city_sqft[city_sqft["count"] >= 30]
    .sort_values("median", ascending=False)
)
```

# Resumen ejecutivo

Este proyecto analiza un conjunto de datos inmobiliarios compuesto inicialmente por 4.600 viviendas de la zona de Seattle. El objetivo es identificar los principales factores relacionados con el precio de las viviendas y detectar diferencias entre mercados según sus características, ubicación y servicios.

Para ello se realizó un proceso de análisis exploratorio de datos (EDA), incluyendo la revisión de valores nulos, duplicados, inconsistencias, valores atípicos, análisis univariante y bivariante y estudio de correlaciones entre variables numéricas.

Tras el proceso de limpieza se obtuvo un conjunto final de 4.354 viviendas y 18 variables, sin valores nulos ni registros duplicados.



## Perfil del mercado inmobiliario

El precio mediano de las viviendas analizadas es de 462.000, mientras que el del precio medio asciende a 547.179,57 $.

La superficie habitable (sqft_living) presenta una mediana de 1.952 sqft, con una media de 2.113,93 sqft.

La distribución de los precios muestra una diferencia importante entre la media y la mediana, debido a la presencia de viviendas de precio elevado. Estos valores extremos fueron revisados durante el proceso de limpieza y se conservaron aquellos registros que presentaban características inmobiliarias coherentes con su precio.

## Factores relacionados con el precio

La superficie habitable es la variable numérica que presenta la mayor correlación con el precio, con un coeficiente de 0,698.

También se observan relaciones positivas entre el precio y:
 
superficie sobre rasante (sqft_above): 0,595
baños (bathrooms): 0,526
vistas (view): 0,383
superficie del sótano (sqft_basement): 0,348
dormitorios (bedrooms): 0,335

Por el contrario, variables como el año de construcción presentan una relación lineal prácticamente inexistente con el precio (0,017).

Estos resultados indican que las características relacionadas con el tamaño y las prestaciones de la vivienda tienen una mayor relación con el precio que la antigüedad del inmueble.

## Ubicación y precio

La ubicación muestra diferencias importantes en el mercado.

Entre las ciudades con al menos 30 viviendas analizadas, Mercer Island presenta el precio mediano más elevado, con 950.833, seguida de Bellevue, con 720.000, y Sammamish, con 658.250

En el extremo inferior se encuentran Covington, con una mediana de 249.500, Des Moines, con 256.000, y Federal Way, con 264.250.

El análisis del precio por sqft permite observar estas diferencias teniendo en cuenta también el tamaño de las viviendas. Mercer Island presenta una mediana de 349,42 dolar/sqft, frente a 146,56 dolar/sqft en Federal Way.

## Características premium

Las viviendas con acceso directo al waterfront presentan precios considerablemente superiores.

La mediana del precio es: 460.000 para viviendas sin waterfront, 995.000 para viviendas con waterfront.

También se observa una clara progresión del precio según la calidad de las vistas. La mediana aumenta desde 440.325 en view = 0 hasta 1.175.000 en view = 4.

Estos resultados muestran una asociación entre determinadas características premium y precios más elevados. Sin embargo, el análisis es descriptivo y no permite establecer relaciones causales.

# Principales conclusiones

Los resultados obtenidos permiten identificar varios patrones relevantes en el mercado inmobiliario analizado:

La superficie habitable es el factor numérico con mayor relación con el precio.
La ubicación genera diferencias importantes en los precios de las viviendas.
Las propiedades con waterfront presentan una mediana de precio muy superior.
Una mejor vista está asociada a precios medianos más elevados.
El número de baños y dormitorios también presenta una relación positiva con el precio.
El precio por sqft permite comparar de forma más adecuada el valor relativo de las viviendas entre ciudades.
El año de construcción presenta una relación prácticamente inexistente con el precio en este conjunto de datos.
